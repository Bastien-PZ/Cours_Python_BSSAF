## Exercices sur Pandas

```python
# Import des données
import pandas as pd
df = pd.read_parquet('/content/FD_NAIS_2024.parquet')
```

```python
df
```

```python
# le mois pour lesquels le nombre de naissances est le plus élevé, pour l'ensemble de la France
print(df.groupby(['MNAIS']).size())

mois_max = df["MNAIS"].value_counts().idxmax()
mois_max
```

```python
# puis, par département, indiquer s'il s'agit du même mois
mois_par_dep = (
    df
    .groupby("DEPNAIS")["MNAIS"]
    .apply(lambda x: x.value_counts().idxmax())
    .reset_index(name="MNAIS")
)

mois_par_dep["MNAIS_07"] = mois_par_dep["MNAIS"].eq("07").map({True: "oui", False: "non"})
print(mois_par_dep.to_string())
```

```python
# la part des naissances gémellaires par département
print(df.groupby('DEPNAIS')['NBENF'].value_counts(normalize=True).unstack().to_string())
```

```python
# enfin, l'âge médian des mères
df["AGEMERE"] = pd.to_numeric(df["AGEMERE"], errors="coerce")
age_median = df["AGEMERE"].median()
print(age_median)
```
