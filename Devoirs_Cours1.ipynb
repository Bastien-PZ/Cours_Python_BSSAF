## Exercice 5

```python
liste = [1,2,3]

def somme_liste(liste):
  somme = 0
  for i in liste:
    somme += i
  return(somme)

somme_liste(liste)
```

## Exercice 7

```python
nombre = 18
liste = []

for i in range(1, nombre+1, 1):
  if nombre % i == 0:
    liste.append(i)

print(liste)
```

## Exercice 8

```python
liste = [1,2,3]

def moyenne_liste(liste):
  somme = 0
  for i in liste:
    somme += i
  return(somme/len(liste))

moyenne_liste(liste)
```

## Exercice 10

```python
liste = [10,20,30,40,50]
nombre = 30

def position_nombre(liste, nombre):
  compteur = 1
  for i in liste:
     if nombre == i:
        res =  compteur
        break
     else:
        compteur += 1
        res = "-1"
  return res

position_nombre(liste, nombre)
```
